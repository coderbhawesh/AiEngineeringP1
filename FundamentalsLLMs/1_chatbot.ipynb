```python
def stateless_reply(prompt: str) -> str:
    return f"Some llm response {prompt}"
    
```

```python
print(stateless_reply("How are you!"))
```

```python
from openai import OpenAI

client = OpenAI(
    base_url="http://127.0.0.1:5001/gateway/mlflow/v1",
    api_key="Not needed",  # API key not needed, configured server-side
)

messages = [{"role": "user", "content": "How are you, chatbot?"}]

response = client.chat.completions.create(
    model="workshop-gemini",  # Endpoint name as model
    messages=messages,
)
print(response.choices[0].message.content)
```
